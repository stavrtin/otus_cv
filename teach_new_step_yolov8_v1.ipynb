```python
from ultralytics import YOLO

# 1. Загружаем модель
model = YOLO("best_800_1200.pt")

# 2. Запускаем обучение
# Внимание: batch=2, так как на RTX 3050 не влезает больше при 960px
results = model.train(
    data='data.yaml',       # Файл с путями к train/val
    epochs=150,
    imgsz=960,              # Разрешение под ваши фото
    batch=2,                # Маленький батч из-за памяти
    amp=True,               # Ускорение через Automatic Mixed Precision
    optimizer='SGD',
    lr0=0.001,              # Скорость обучения
    patience=50,            # Ранняя остановка
    workers=2,              # Для Windows
    name='train_1200_1600'  # Название папки с результатами
)
```
